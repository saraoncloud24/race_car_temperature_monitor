```python
import matplotlib.pyplot as plt
#fake temperature data that measures temperature every 10 seconds
times = [0, 10, 20, 30, 40, 50]
temperatures = [42, 51, 63, 76, 83, 88]
warningTemp = 80

#print each time, temp, and status
for i in range(len(times)):
  if temperatures[i]>= warningTemp:
    status = "WARNING"
  else:
    status = "OK"
  print(f"Time: {times[i]} sec | Temp: {temperatures[i]} C | Status: {status}")

#plot the temperature
plt.plot(times, temperatures, marker="o")
plt.axhline(y=warningTemp, linestyle="--", label="Warning Threshold")
plt.xlabel("Time (seconds)")
plt.ylabel("Temperature (C)")
plt.title("Race Car Temperature Monitoring")
plt.legend()
plt.show()

```
